```python
#DATA CLEANING
#Check for duplicate rows
#Are they true duplicates?
df[df.duplicated(keep=False)]
```

```python
#Removing duplicates
df = df.drop_duplicates().reset_index(drop=True)

#check if any rows are remaining after cleaning
df.duplicated().sum()
```

```python
#Fixing MRR Column
df["MRR_Raw"] = df["MRR"]

df["MRR"] = (
    df["MRR"]
    .astype(str)
    .str.replace(r"[^0-9.]", "", regex=True)
    .astype(float)
)
```

```python
df["MRR"]
```

```python
df["Signup Date"].head()
```

```python
#Cleaning Mixed-Up Date Formats
DATE_COLS = ["Signup Date", "Last Login Date", "Churn Date"]

for col in DATE_COLS:
    df[f"{col}_Raw"] = df[col]

    df[col] = pd.to_datetime(
        df[col],
        format="mixed",
        dayfirst=False,
        errors="coerce"
    )
```

```python
df[DATE_COLS]
```

```python
#Plan names
df["Plan"].value_counts(dropna=False)
```

```python
#Standardise plan names
# Clean
df["Plan"] = (
    df["Plan"]
    .str.strip()
    .str.replace(" Plan", "", regex=False)
    .str.title()
)

#Verify
df["Plan"].value_counts()
```

```python
#Region names
df["Region"].value_counts(dropna=False)
```

```python
# Cleaning
df["Region"] = (
    df["Region"]
    .str.replace("_", " ", regex=False)
    .str.replace(" Region", "", regex=False)
    .str.strip()
    .str.title()
)

# Verifying
df["Region"].unique()
```

```python
#Checking for missing values
df.isnull().sum()
```

```python
#Replacing 3 categorical columns that have nulls with unknown 
categorical_cols = [
    "Industry",
    "Acquisition Channel",
    "Sales Rep"
]

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")
```

```python
#Dealing with null sales reps
df["Sales Rep"] = df["Sales Rep"].replace("N/A", "Unknown")
```

```python
#Flagging missing NPS
df["NPS_Missing_Flag"] = df["NPS Score"].isnull().astype(int)
```

```python
#Failed Payment Retry
df["Failed Payment Retry"].value_counts(dropna=False)
```

```python
#Fixing failed payment retry
#nulls mean the customer likely didn’t have a failed payment, not that data is missing
df["Failed Payment Retry"] = df["Failed Payment Retry"].fillna("No Failed Payment")
```

```python
#Validating MRR vs ARR
#ARR should equal monthly recurring revenue × 12
df["Expected_ARR"] = df["MRR"] * 12

df["ARR_Difference"] = df["ARR"] - df["Expected_ARR"]

df[df["ARR_Difference"].abs() > 0.01]
```

```python
#Clean validation flag
df["ARR_Valid_Flag"] = (df["ARR_Difference"].abs() <= 0.01).astype(int)
```

```python
df["ARR_Valid_Flag"]
```

```python
#Add days since last login
reference_date = df["Last Login Date"].max()

df["Days_Since_Login"] = (
    reference_date - df["Last Login Date"]
).dt.days
```

```python
df["Days_Since_Login"]
```

```python
#Adding useful Power BI flags
df["Is_Churned"] = df["Churn Date"].notna().astype(int)

df["Low_Utilization_Flag"] = (
    df["Seat Utilization Pct"] < 50
).astype(int)

df["High_Support_Flag"] = (
    df["Support Tickets"] >= 8
).astype(int)

df["Inactive_60_Days_Flag"] = (
    df["Days_Since_Login"] > 60
).astype(int)
```
